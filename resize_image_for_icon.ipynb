```python
from PIL import Image

import numpy as np
import pandas as pd
```

```python
# load image 
im_path = r'./img_purim_2026/little_gizmo.jpg'

# resize image to 180x180, 32x32, and 16x16 and save each as png

def resize_and_save(im_path, w, h);
    im = Image.open(im_path)
    new_size = (h, w)

    resize_im = im.resize(new_size, Image.Resampling.LANCZOS)
    resize_im.save(fr'./img_resize{w}x{h}.png')
```
